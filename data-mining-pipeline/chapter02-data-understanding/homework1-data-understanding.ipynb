## Problem 1 — descriptive statistics

Chapter 2 describes numeric attributes with measures of center and spread. The mean and median summarize the center; standard deviation summarizes spread; quartiles and IQR describe the middle half of the observations. This implementation converts the selected column to numeric values and excludes missing or non-numeric entries. `Series.std()` uses pandas' sample standard deviation by default (`ddof=1`), and `quantile()` uses pandas' default interpolation.

```python
def calculate(dataFile, col_num):
    """Return descriptive statistics for the selected numeric column."""
    numObj, minValue, maxValue, mean, stdev, Q1, median, Q3, IQR = [
        0, "inf", "-inf", 0, 0, 0, 0, 0, 0
    ]

    data = pd.read_csv(dataFile)
    values = pd.to_numeric(data.iloc[:, col_num], errors="coerce").dropna()

    numObj = len(values)
    if numObj > 0:
        minValue = values.min()
        maxValue = values.max()
        mean = values.mean()
        stdev = values.std()
        Q1 = values.quantile(0.25)
        median = values.quantile(0.50)
        Q3 = values.quantile(0.75)
        IQR = Q3 - Q1

    return numObj, minValue, maxValue, mean, stdev, Q1, median, Q3, IQR
```
