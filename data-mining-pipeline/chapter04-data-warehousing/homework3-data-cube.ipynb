This assignment applies Chapter 4 concepts: aggregating measures across dimensions and representing the results as a multidimensional data cube.

## Chapter 4 concepts used here

A data cube summarizes a measure (here, `sales`) across dimensions such as time, geography, and product category. Grouping sales by year, country, or category produces lower-dimensional summaries (cuboids). Grouping simultaneously by year and category gives the coordinates for a three-dimensional visualization: year and category are the dimensions, and total sales is the measure.

The assignment's year/category plot is a small OLAP-style view. It supports comparing sales across time and product groups; rolling up or slicing the cube would provide other summary views.

## Assignment description

The dataset contains daily product sales for five years across retail stores. Its attributes are `year`, `month`, `day`, `country`, `state`, `city`, `store`, `category`, `product`, and `sales`. The relative path is `./dataset/dataset.csv`.

Section 1 asks for total sales grouped separately by year, country, and category. Each function returns a Python dictionary whose keys are the grouping values and whose values are summed sales. Section 2 asks for Python lists of year (`x`), category (`y`), and total sales (`z`) aggregated by year and category, for plotting as a 3D data cube.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # required by the assignment environment
```

## Section 1 — aggregate sales by one dimension

Each function reads the supplied dataset and groups the `sales` measure by one dimension. `sum()` performs the aggregation, and `to_dict()` returns the required key-to-total mapping.

```python
def aggregate_by_year():
    # Reading dataset
    df = pd.read_csv('./dataset/dataset.csv')
    d = df.groupby('year')['sales'].sum().to_dict()
    return d
```

```python
def aggregate_by_country():
    # Reading dataset
    df = pd.read_csv('./dataset/dataset.csv')
    d = df.groupby('country')['sales'].sum().to_dict()
    return d
```

```python
def aggregate_by_category():
    # Reading dataset
    df = pd.read_csv('./dataset/dataset.csv')
    d = df.groupby('category')['sales'].sum().to_dict()
    return d
```

## Section 2 — prepare the data cube coordinates

The cube summarizes total sales for each `(year, category)` pair. The function returns three aligned Python lists: `x` contains years, `y` contains category names, and `z` contains the corresponding summed sales. The assignment plots category codes on the vertical coordinate while labeling them with category names.

```python
def plot_data_cube():
    '''Output: x, y, z as Python lists.'''
    df = pd.read_csv('./dataset/dataset.csv')
    grouped = df.groupby(['year', 'category'], as_index=False)['sales'].sum()

    x = list(grouped['year'])
    y = list(grouped['category'])
    z = list(grouped['sales'])
    return x, y, z
```

## Plot the data cube

This reproduces the assignment's visualization: year on the x-axis, category on the y-axis, and aggregated sales on the z-axis.

```python
x, y, z = plot_data_cube()
y_codes = list(pd.Categorical(y).codes)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(projection='3d')
ax.scatter(x, y_codes, z)

plt.xticks(np.arange(min(x), max(x) + 1, 1))
plt.yticks(np.arange(min(y_codes), max(y_codes) + 1, 1))
ax.set_yticklabels(pd.Categorical(y).categories)
ax.set_xlabel('Year')
ax.set_ylabel('Category')
ax.set_zlabel('TotalSales')
ax.set_title('Datacube: Aggregate Sales across year and category')
plt.show()
```
